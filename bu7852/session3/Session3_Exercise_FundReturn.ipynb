```python
# --- The fund's historical annual returns (%) -----------------------
# 200 simulated hypothetical years, same fund (mean ~8%, sd ~11.8%).
import numpy as np, matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(3)
fund_returns = rng.normal(8.0, 11.8, size=200)
THIS_YEAR = -18
print(f"History: mean {fund_returns.mean():.1f}%   sd {fund_returns.std(ddof=1):.1f}%")
print(f"This year: {THIS_YEAR}%")
```

## Step 1, check normality first: the Q-Q plot

Before you trust any normal-curve tool (z-score, empirical rule), check it applies. Does the fund's return history look normal?

```python
stats.probplot(fund_returns, dist='norm', plot=plt.gca())
plt.title('Is the fund\'s return history normal?')
plt.show()
```

## Step 2, the z-score: how unusual is -18%?

Compute the z-score for this year's return against the fund's own history.

```python
mu, sd = fund_returns.mean(), fund_returns.std(ddof=1)
z = (THIS_YEAR - mu) / sd
print(f"z = ({THIS_YEAR} - {mu:.1f}) / {sd:.1f} = {z:.2f}")
```

## Step 3, the empirical rule and the exact probability

Use the empirical rule (68-95-99.7) for a quick read, then the exact normal-curve probability for the precise one. Do they agree?

```python
# Exact probability of a year this bad or worse, assuming normality:
p_this_bad_or_worse = stats.norm.cdf(THIS_YEAR, mu, sd)
print(f"P(return <= {THIS_YEAR}%) = {p_this_bad_or_worse:.4f} (about 1 in {1/p_this_bad_or_worse:.0f} years)")
```

## Step 3b, optional stretch: benchmark against peers (NAV benchmarking)

So far you have asked "was -18% unusual *for this fund*?" The audit move from last week asks the sharper question: "was it unusual *compared to its peers*?" Benchmarking a unit against its peer group's median is how an auditor spots the one that does not fit.

Here is the same year for 40 comparable funds. Judge -18% two ways: the gap to the **peer median**, and where it sits in the **peer distribution** (a z-score or percentile). The two lenses, own-history and peers, can disagree, and that gap is the finding.

```python
# --- 40 comparable funds, this same year (%) ------------------------
# A broadly down year for the sector, so peers are off too, but by how much?
rng_p = np.random.default_rng(21)
peers = rng_p.normal(-5.0, 8.0, size=40)
peer_median = np.median(peers)
peer_mu, peer_sd = peers.mean(), peers.std(ddof=1)

gap = THIS_YEAR - peer_median
percentile = (peers <= THIS_YEAR).mean() * 100
z_peer = (THIS_YEAR - peer_mu) / peer_sd

print(f"peer median: {peer_median:.1f}%     our fund: {THIS_YEAR}%     gap: {gap:+.1f} pp")
print(f"our fund sits at about the {percentile:.0f}th percentile of its peers")
print(f"z vs peers: {z_peer:.2f}   (against its own history it was z = {z:.2f})")

plt.figure(figsize=(8, 4))
plt.hist(peers, bins=12, color='#0D73B8', alpha=0.8)
plt.axvline(peer_median, color='#3E8E5A', lw=2, label=f'peer median {peer_median:.1f}%')
plt.axvline(THIS_YEAR, color='#C0392B', lw=2, label=f'our fund {THIS_YEAR}%')
plt.xlabel("this year's return (%)"); plt.legend()
plt.title('Our fund against 40 peers this year'); plt.show()
```
