```python
# --- The two portfolios: historical monthly returns (%) ------------
import numpy as np, matplotlib.pyplot as plt
X = np.array([2.5, 3.1, 2.8, 2.9, 3.0, 2.7, 3.2, 2.6, 2.9, 3.3])   # Portfolio X, Conservative Growth
Y = np.array([4.0, 1.5, 5.0, 0.5, 6.0, 2.0, 7.0, 1.0, 8.0, 3.0])   # Portfolio Y, Dynamic Opportunity
print('X:', X)
print('Y:', Y)
```

## Step 1, put numbers on each portfolio

For **each** portfolio compute the **mean**, the **standard deviation**, and the **coefficient of variation (CV = SD / mean)**. The CV is the key one here: it is risk per unit of return, so it compares consistency fairly even when the averages differ. Fill in the cell, or ask Claude and then *check its working*.

```python
for name, p in [('Portfolio X', X), ('Portfolio Y', Y)]:
    m = p.mean()
    s = p.std(ddof=1)          # sample standard deviation (n - 1)
    cv = s / m * 100
    print(f'{name}:  mean {m:4.2f}%   sd {s:4.2f}   CV {cv:5.1f}%')
```

## Step 2, see it: box plots side by side

A box plot shows the spread and the consistency at a glance. Plot both portfolios on the same axis and compare the box widths.

```python
plt.figure(figsize=(7, 4))
plt.boxplot([X, Y], orientation='vertical', widths=0.55,
            tick_labels=['Portfolio X\n(conservative)', 'Portfolio Y\n(dynamic)'])
plt.ylabel('monthly return (%)')
plt.title('Which one is more consistent?')
plt.show()
```
