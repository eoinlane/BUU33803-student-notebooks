Probability is where the long-run frequency settles. This notebook flips a fair coin thousands of
times and plots the running proportion of heads, so you can watch it settle.

**How to run it:** click *Runtime > Run all*. Then change the numbers in the cells and run again.

## 1. Flip one coin 3,000 times

```
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)      # fixed seed: the same flips every run
n_flips = 3000
p_heads = 0.5                        # try 0.55 later

flips = (rng.random(n_flips) < p_heads).astype(int)       # 1 = heads, 0 = tails
flip_number = np.arange(1, n_flips + 1)
running_proportion = np.cumsum(flips) / flip_number       # heads so far / flips so far

for n in [10, 100, 1000, 3000]:
    print(f"after {n:>4} flips: proportion of heads = {running_proportion[n - 1]:.3f}")
```

## 2. The graph

```
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(flip_number, running_proportion, color="#1F497D", lw=1.5)
ax.axhline(0.5, color="#C0392B", ls="--", lw=2, label="true probability = 0.5")
ax.set_xscale("log")
ax.set_ylim(0, 1)
ax.set_xlabel("number of flips (log scale)")
ax.set_ylabel("proportion of heads so far")
ax.set_title("One coin: the running proportion of heads")
ax.legend()
plt.show()
```

Early on the line jumps around, because one flip moves the average a lot. Later each flip
barely moves it. That settling is the law of large numbers.

## 3. Six coins at once

```
fig, ax = plt.subplots(figsize=(9, 4.5))
for coin in range(6):
    f = (rng.random(n_flips) < p_heads).astype(int)
    ax.plot(flip_number, np.cumsum(f) / flip_number, color="#1F497D", alpha=0.45, lw=1.3)
ax.axhline(0.5, color="#C0392B", ls="--", lw=2, label="true probability = 0.5")
ax.axvspan(1, 30, color="#E8A33D", alpha=0.15)      # the short, noisy zone
ax.set_xscale("log")
ax.set_ylim(0, 1)
ax.set_xlabel("number of flips (log scale)")
ax.set_ylabel("proportion of heads so far")
ax.set_title("Six coins: wild at first, then they all settle on 0.5")
ax.legend(loc="lower right")
plt.show()
```
