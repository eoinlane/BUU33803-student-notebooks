Here's the kind of code it will give you, run it:

```python
import matplotlib.pyplot as plt

plt.scatter(pizza['student_population'], pizza['annual_sales'])
plt.xlabel('Student population nearby (000s)')
plt.ylabel('Annual sales ($000s)')
plt.title("Armand's Pizza, sales vs nearby students")
plt.show()

pizza.describe()
```

## Step 3, Build the model  (let Claude drive)

Now fit the line. A good prompt to Claude:

> *"Fit a simple linear regression of `annual_sales` on `student_population` in Python using statsmodels, and print the full summary so I can read the coefficient, its p-value, and R²."*

Run the code it gives you (this is the standard recipe):

```python
import statsmodels.formula.api as smf

model = smf.ols('annual_sales ~ student_population', data=pizza).fit()
print(model.summary())
```

## Step 4, Reveal & reflect

Now **read the output** (the three numbers from the lecture). In the summary, find:

1. **The slope**, the `coef` on `student_population`. What sign? Roughly what size?
2. **Significance**, its `P>|t|`. Real effect, or could it be chance?
3. **R²**, the `R-squared` line. How much of the variation does the model explain?

**Were you right?** Compare each to the guess you committed to in Step 1. What surprised you?

<details><summary>✅ Stuck? What you should find</summary>

Slope ≈ **5**, intercept ≈ **60**, R² ≈ **0.90**, p-value ≈ **0.00003** (very significant). In plain English: **each extra 1,000 students nearby is worth about \$5,000 in annual sales**, and the relationship is strong and almost certainly real, not luck.

</details>

## Step 5, Turn it into a business decision

A model is only useful if it changes a decision. **Armand's is scouting a new location with 18,000 students nearby.** Use your model to predict its sales, then decide.

> Ask Claude: *"Using this model, what annual sales would you predict for a location with 18,000 students nearby, and how confident should I be in that single prediction?"*

```python
# 18,000 students = student_population of 18 (the column is in thousands)
pred = model.predict({'student_population': [18]})
print('Predicted annual sales ($000s):', round(float(pred.iloc[0]), 1))
```
