> **The data:** Camac Drinks, a fictional sparkling-water brand, advertises in 200 towns across Ireland and Britain. For each town: spend on `TV`, `Radio`, `Newspaper` (in thousands of euros) and the resulting `Sales` (in thousands of cans).
>
> *Where it really comes from:* a teaching dataset, a variant of the Advertising data in *An Introduction to Statistical Learning* (the spend columns match the book, the sales figures don't). Trust this file, not remembered textbook numbers.

```python
# --- Load the data -------------------------------------------------
# The data loads from the course's public notebook repo, nothing to upload.
# No internet? Set DATA_URL = "" and upload the file when prompted (Colab only).
DATA_URL = "https://raw.githubusercontent.com/eoinlane/BUU33803-student-notebooks/main/session1/data/advertising.csv"

import pandas as pd

if DATA_URL:
    ads = pd.read_csv(DATA_URL)
else:
    try:
        from google.colab import files
        print("Upload advertising.csv …")
        up = files.upload()
        ads = pd.read_csv(next(iter(up)))
    except ImportError:
        ads = pd.read_csv("advertising.csv")   # local fallback (repo: data/advertising.csv)

ads.head()
```

## Step 1, Meet the data *before* you analyse it

First question to Claude is never "build me a model." It's "what am I even looking at?"

```python
ads.shape          # how many rows and columns?
ads.describe()      # ranges, averages, does anything look odd?
```

## Step 2, Predict before you compute  🎯

**Ask the room, out loud, before running anything:** which channel do you think drives sales most, TV, Radio, or Newspaper? Get them to *commit*. We'll see who's right.

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 3, figsize=(13, 4), sharey=True)
for a, col in zip(ax, ['TV', 'Radio', 'Newspaper']):
    a.scatter(ads[col], ads['Sales'], alpha=0.6)
    a.set_xlabel(col + ' spend'); a.set_ylabel('Sales')
plt.tight_layout(); plt.show()
```

## Step 3, The plant-a-mistake beat  ⚠️

Now **deliberately invite Claude to over-claim**, then catch it. For example, ask Claude:

> *"Newspaper spend and Sales are positively correlated, so newspaper ads cause sales, right?"*

Claude (or a careless analyst) might agree. **Catch it out loud:** correlation isn't causation, and `Newspaper` may just ride along with `Radio` (a confounder). Show the correlation matrix:

```python
ads.corr(numeric_only=True).round(2)
```

Point out: `Newspaper`'s correlation with `Sales` is real but weak, and partly explained by its correlation with `Radio`. **This is the whole game:** Claude will happily produce a confident story; your job is to check whether the data actually supports it.

## Step 4, Good question vs. weak question

| Weak prompt | Strong prompt |
|---|---|
| "analyse this data" | "which channel gives the most *sales per euro spent*?" |
| "make a chart" | "show TV vs Sales and tell me if the relationship looks linear" |
| "is this good?" | "what would make you *distrust* a model built on this data?" |

The precise, business-framed question is the one that earns a useful answer.
