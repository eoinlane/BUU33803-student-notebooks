## What the columns mean

Each row is **one day** (731 days, 2011–2012).

| Column | Meaning |
|---|---|
| `season` | 1 spring · 2 summer · 3 fall · 4 winter |
| `yr` | 0 = 2011, 1 = 2012 |
| `mnth` | month, 1–12 |
| `holiday` | 1 if a holiday |
| `weekday` | day of week (0 = Sunday) |
| `workingday` | 1 if a normal working day (not weekend/holiday) |
| `weathersit` | 1 clear · 2 misty · 3 light rain/snow · 4 heavy weather |
| `temp` | temperature, **normalised 0–1** (÷41 °C) |
| `atemp` | "feels-like" temperature, normalised 0–1 (÷50 °C) |
| `hum` | humidity, normalised 0–1 |
| `windspeed` | wind speed, normalised 0–1 |
| `casual` | casual (non-member) riders that day |
| `registered` | registered (member) riders that day |
| `cnt` | **total riders that day** (casual + registered) |
